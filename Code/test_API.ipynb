# Flugpreis-Check API testen
Lokal: API zuerst starten mit `uvicorn Code.main:app --reload`.
Cloud: `URL` auf die render.com-Adresse umstellen.

```python
import requests

URL = "http://127.0.0.1:8000"   # Cloud: "https://<dein-service>.onrender.com"
```

```python
flug = {
    "airline": "Indigo",
    "von": "Delhi",
    "nach": "Mumbai",
    "abflugzeit": "Evening",
    "stopps": "zero",
    "klasse": "Economy",
    "tage_bis_abflug": 45,
}

resp = requests.post(f"{URL}/flugpreis-check", json=flug)
data = resp.json()

print(f"Preis heute:  CHF {data['preis_heute_chf']}")
print(f"Empfehlung:   {data['empfehlung']}")
print(f"Bester Preis: CHF {data['bester_preis_chf']} ({data['bester_zeitpunkt_tage_vor_abflug']} Tage vor Abflug)")
print(f"Ersparnis:    CHF {data['ersparnis_chf']}")
```

```python
# Preisverlauf: Wie entwickelt sich der Preis bis zum Abflug?
for p in data["preisverlauf"][::5]:
    print(f"{p['tage_bis_abflug']:>2} Tage vorher: CHF {p['preis_chf']}")
```
